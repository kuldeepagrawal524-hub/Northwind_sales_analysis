```python
import pandas as pd

query = """
SELECT
    order_id,
    order_value
FROM (
    SELECT
        order_id,
        SUM(quantity * unit_price * (1 - discount)) AS order_value,
        NTILE(10) OVER (
            ORDER BY SUM(quantity * unit_price * (1 - discount)) DESC
        ) AS bucket
    FROM orders_detail
    GROUP BY order_id
) t
WHERE bucket = 1
ORDER BY order_value DESC;
"""

df = pd.read_sql(query, conn)
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# SQL query
query = """
SELECT
    EXTRACT(MONTH FROM o.order_date) AS month,
    SUM(od.quantity * od.unit_price * (1 - od.discount)) AS total_sales
FROM orders o
JOIN orders_detail od
    ON o.order_id = od.order_id
GROUP BY month
ORDER BY month;
"""

df = pd.read_sql(query, conn)

print(df)   # debug ke liye

# Heatmap ke liye reshape
heatmap_data = df.pivot_table(
    values='total_sales',
    index=['month']
)

plt.figure(figsize=(8, 4))
plt.imshow(heatmap_data, aspect='auto')
plt.colorbar(label='Total Revenue')

plt.yticks(
    ticks=range(len(heatmap_data.index)),
    labels=heatmap_data.index
)

plt.xticks([0], ['Revenue'])

plt.xlabel("Sales")
plt.ylabel("Month")
plt.title(" Month-wise Revenue Heatmap")

plt.tight_layout()
plt.show()
```
